```python
# Introduction to Pandas Library

"""
Pandas stands for "Panel Data".

It is a powerful Python library used for:

1. Working with structured data
2. (EDA)
3. Data cleaning and preprocessing
4. Data manipulation and transformation
"""
```

```python
import pandas as pd

# -------------------------------
# Create a Series
# -------------------------------
print("Pandas Series:")
s = pd.Series([10, 20, 30, 40])
display(s)


# -------------------------------
# Create a DataFrame
# -------------------------------
print("\nDataFrame:")
data = {
    "Name": ["a", "b", "c"],
    "Marks": [85, 90, 78]
}

df = pd.DataFrame(data)
display(df)


print("\nShape of DataFrame (Rows, Columns):")
print(df.shape)

print("\nNumber of Dimensions:")
print(df.ndim)

print("\nTotal Number of Elements:")
print(df.size)

print("\nColumn Names:")
print(df.columns)

print("\nData Types of Each Column:")
print(df.dtypes)

print("\nRow and Column Axis Labels:")
print(df.axes)




print("\nData in 'Name' Column:")
display(df["Name"])

print("\nRow with index label 1:")
display(df.loc[1])

print("\nValue at row index 1 and column index 1:")
print(df.iloc[1, 1])



print("\nFirst 5 Rows:")
display(df.head())

print("\nLast 5 Rows:")
display(df.tail())

# Add a new column


'''If marks 80-90-First class
IF marks >=90-distinction
If marks >80-Second class'''

df["Grade"]=["First class","distinction","Second class"]
df
print()


# Change marks of student A to 88
df.iloc[0,1] = 81
df
print()

```

```python
# Update Data in DataFrame
# Change marks of student A to 88

df.loc[0,"Marks"] = 81
df
print()


```

```python
# Delete the 'Grade' column
df.drop("Grade", axis=1, inplace=True, errors='ignore')

df
```

```python
# Perform Basic Statistical Analysis

mean1=df["Marks"].mean()
display(mean1)
print()

min1=df["Marks"].min()
display(min1)
print()

min1=df["Marks"].max()
display(min1)
print()



# perform basic Statistical Analysis
```

```python
# apply condition (filtering)
# display student scoring more than 80

df[df["Marks"] > 80]
```
